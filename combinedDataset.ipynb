```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
dataset1 = pd.read_csv('Dataset/Year/natality2020ps.csv')
dataset2 = pd.read_csv('Dataset/Year/natality2021ps.csv')
dataset3 = pd.read_csv('Dataset/Year/natality2022ps.csv')
dataset4 = pd.read_csv('Dataset/Year/natality2023ps.csv')
dataset5 = pd.read_csv('Dataset/Year/natality2024ps.csv')
```

```python
# Rename 'f_bfacil' to 'f_facility' in datasets 3, 4, 5
dataset3 = dataset3.rename(columns={'f_bfacil': 'f_facility'})
dataset4 = dataset4.rename(columns={'f_bfacil': 'f_facility'})
dataset5 = dataset5.rename(columns={'f_bfacil': 'f_facility'})
```

```python
dataset = pd.concat([dataset1, dataset2, dataset3, dataset4, dataset5], ignore_index=False)
dataset.shape
```

```python
set1 = set(dataset1.columns)
set2 = set(dataset2.columns)
set3 = set(dataset3.columns)
set4 = set(dataset4.columns)
set5 = set(dataset5.columns)

print("1 vs 2:", set1 == set2)
print("1 vs 3:", set1 == set3)
print("1 vs 4:", set1 == set4)
print("1 vs 5:", set1 == set5)

print("Columns in dataset1 but not in dataset2:", set1 - set2)
print("Columns in dataset2 but not in dataset1:", set2 - set1)

print("Columns in dataset1 but not in dataset3:", set1 - set3)
print("Columns in dataset3 but not in dataset1:", set3 - set1)

print("Columns in dataset1 but not in dataset4:", set1 - set4)
print("Columns in dataset4 but not in dataset1:", set4 - set1)

print("Columns in dataset1 but not in dataset5:", set1 - set5)
print("Columns in dataset5 but not in dataset1:", set5 - set1)
```
